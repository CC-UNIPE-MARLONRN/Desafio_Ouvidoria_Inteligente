# Chunking de Manifestações Longas

Para textos longos com mais de 500 caracteres, o chunking divide a manifestação em segmentos com menor perda de contexto. A estratégia `RecursiveCharacterTextSplitter` ajuda a manter o sentido pela preservação de fronteiras semânticas como parágrafos e frases.

Quando o overlap aumenta, há maior continuidade entre chunks consecutivos e menos fragmentação de ideias. Porém, se o overlap for exagerado, há duplicação de informação e perda de clareza na separação dos blocos. A configuração mais equilibrada para a base foi chunk_size=220 e chunk_overlap=40.


```python
from ouvidoria_utils import chunk_dataframe, get_longest_manifestations, load_manifestacoes
df = load_manifestacoes('manifestacoes.json')
long_df = get_longest_manifestations(df, n=5)
chunks = chunk_dataframe(long_df, chunk_size=220, chunk_overlap=40)
print(chunks.head(10).to_markdown(index=False))
```

## Visualização e reflexão

Os chunks de uma mesma manifestação tendem a ficar próximos no espaço semântico quando o overlap preserva continuidade temática e os trechos são coesivos. Em contraste, altas variações de tópico dentro do mesmo texto podem separar os blocos em regiões diferentes do espaço, especialmente quando há mudanças de assunto.

A configuração escolhida preserva melhor o sentido das denúncias porque evita que o início de um tópico seja perdido no final do chunk anterior, ao mesmo tempo em que não repete conteúdo excessivamente.
