# Deteção de Duplicatas

A função `detectar_duplicatas(textos, limiar=0.85)` verifica pares com similaridade de cosseno acima do limite escolhido. O limiar foi fixado em 0.85 para reduzir falsos positivos, mas ainda captar duplicatas semânticas com variações léxicas.


```python
import pandas as pd
from ouvidoria_utils import detectar_duplicatas, load_manifestacoes
df = load_manifestacoes('manifestacoes.json')
pares = detectar_duplicatas(df['texto'].tolist(), limiar=0.85)
print(pd.DataFrame(pares)[['similaridade', 'texto_1', 'texto_2']].head().to_markdown(index=False))
```

## Pares duplicados e análise

Foram detectados 2 pares acima do limiar.

Valores próximos de 0.85 funcionam bem para capturar semelhanças fortes sem classificar temas afins como duplicatas. Em termos práticos, o limite deve ser ajustado com validação manual para balancear falsos positivos e falsos negativos.

Exemplos típicos: 'buraco enorme na Av. Brasil' vs 'asfalto todo esburacado na avenida principal'; 'posto de saúde sem médico' vs 'falta atendimento no PSF'. Ensaios com a mesma base revelam que duplicatas reais de linguagem muito distinta são mais bem detectadas por embeddings do que por BoW.
