# Análise Comparativa de Representações

Este notebook compara BoW, TF-IDF e embeddings para os textos do desafio e discute a eficácia de cada uma para detectar semelhanças entre manifestações.


```python
from pathlib import Path
import pandas as pd
from ouvidoria_utils import compare_representations, load_manifestacoes
df = load_manifestacoes('manifestacoes.json')
pairs = [('M003', 'M017'), ('M008', 'M022'), ('M008', 'M031')]
resultado = compare_representations(df, pairs, model_name='paraphrase-multilingual-MiniLM-L12-v2')
print(resultado.to_markdown(index=False))
```

## Tabela comparativa

| Par | BoW | TF-IDF | Embeddings |
| --- | ---: | ---: | ---: |
| M003 × M017 | 0.0913 | 0.0224 | 0.5664 |
| M008 × M022 | 0.2739 | 0.1400 | 0.6283 |
| M008 × M031 | 0.1000 | 0.0253 | 0.1248 |

Como a semântica da linguagem costuma ser mais rica que o vocabulário literal, os embeddings capturam melhor as nuances entre sinônimos e expressões equivalentes. BoW e TF-IDF tendem a subestimar relações semânticas quando mecanismos de expressão diferentes aparecem, como 'asfalto esburacado' e 'buraco enorme'.

A principal limitação do BoW é sua natureza extremamente lexical; TF-IDF melhora a ponderação, mas ainda depende da sobreposição direta de termos. Os embeddings são mais robustos para linguagem natural, porém exigem maior custo computacional e podem sofrer com ruído quando o texto é curto ou ambíguo.
